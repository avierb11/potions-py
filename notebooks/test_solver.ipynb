```python
import importlib 
import potions as pt
import torch
```

```python
n: int = 1_000_000
fcs = 100 + (1000 - 100) * torch.rand(size=(n,)).float().to("cuda:0")
lps = 0.01 + (1.0 - 0.01) * torch.rand(size=(n,)).float().to("cuda:0")
betas = 0.05 + (5.0 - 0.5) * torch.rand(size=(n,)).float().to("cuda:0")
k0s = 0.01 + (0.1 - 0.01) * torch.rand(size=(n,)).float().to("cuda:0")
thrs = 0.0 * (100 - 0.0) * torch.rand(size=(n,)).float().to("cuda:0")

surf = pt.SurfaceZone(
    fc=fcs,
    lp=lps,
    beta=betas,
    k0=k0s,
    thr=thrs
)
```

```python
d = pt.HydroForcing(
    precip=5.0, temp=10.0, pet=2.5, q_in=5.0,
)
```

```python
x = surf.step(
    s_0=240.0,
    d=d,
    dt=1.0
)
```
